```python
# @title ⚙️ การตั้งค่า — โปรดรันเซลล์นี้ก่อน (ไม่ต้องแก้ไข)
# ============================================================================
# 🎯 วัตถุประสงค์ของเซลล์นี้
#   สร้างตัวแปร REPORT สำหรับเก็บคำตอบจากแบบฟอร์มทุกขั้น และฟังก์ชัน save(...) ที่แบบฟอร์มเรียกใช้
#   เซลล์นี้ไม่โหลดข้อมูลใด ๆ — กลุ่มเป็นผู้เลือกและโหลดชุดข้อมูลของตนเองในพื้นที่ทำงาน
# ============================================================================
# นำเข้าฟังก์ชันสำหรับแสดงข้อความแบบ Markdown ในโน้ตบุ๊ก (ใช้ในรายงานของกลุ่ม)
from IPython.display import display, Markdown

# REPORT : dict สำหรับเก็บคำตอบ โดยกุญแจ (key) คือชื่อขั้น และค่าคือรายการ (หัวข้อ, คำตอบ)
REPORT = {}
# SECTIONS : ลำดับของขั้นในรายงานของกลุ่ม
SECTIONS = ["1. กลุ่มและโครงงาน", "2. Project canvas", "3. ข้อมูล label และ leakage", "4. Baseline และโมเดล",
            "5. การประเมินผลอย่างตรงไปตรงมา", "6. การสาธิต", "7. ข้อจำกัดและเส้นทางสู่การใช้งานจริง",
            "8. การเตรียมการนำเสนอ", "9. การเตรียมตอบคำถามมาตรฐาน 5 ข้อ"]


def filled(v):
    # คืนค่า True เมื่อช่องคำตอบมีค่า ได้แก่ ข้อความที่ไม่ว่าง (และมิใช่ตัวเลือก "— เลือก —") หรือช่องที่ทำเครื่องหมายแล้ว
    if isinstance(v, str):
        return bool(v.strip()) and not v.startswith("—")
    return bool(v)


def save(section, items):
    # บันทึกคำตอบของขั้น section (dict ของหัวข้อและคำตอบ) ลงใน REPORT แล้วแสดงสถานะของแต่ละข้อ
    REPORT[section] = list(items.items())
    done = sum(filled(v) for v in items.values())
    print(f"💾 บันทึกแล้ว: {section} — กรอกครบ {done} จาก {len(items)} ข้อ")
    for label, v in items.items():
        print(f"  {'✅' if filled(v) else '⬜'} {label}")


print("✅ การตั้งค่าเสร็จสมบูรณ์ — ดำเนินการขั้นที่ 1 ต่อได้")
```

**Label เป็นการตัดสินใจเชิงออกแบบ (the label is a design decision)** — โมเดลเรียนรู้ที่จะเลียนแบบ label ดังนั้น การเลือก label จึงต้องรอบคอบไม่น้อยกว่าการเลือกโมเดล

- **Label ที่เป็นตัวแทน (proxy label) มีอคติ:** อัลกอริทึมที่ใช้*ค่าใช้จ่ายด้านสุขภาพ*เป็นตัวแทนของ*ความจำเป็นด้านสุขภาพ* ทำให้ผู้ป่วยผิวดำซึ่งมีความจำเป็นเท่ากันแต่มีค่าใช้จ่ายต่ำกว่า**ได้รับบริการน้อยกว่าที่ควร**
- **รหัสวินิจฉัยเป็นตัวแทนเช่นกัน:** ลงรหัสเพื่อการเบิกจ่าย ไม่ปรากฏเมื่อไม่มีการตรวจ และแตกต่างกันระหว่างโรงพยาบาล
- **ผู้ใดอยู่ในตัวหาร:** ผู้ที่เสียชีวิตหรือย้ายโรงพยาบาลไม่สามารถกลับมารักษาซ้ำได้ กลุ่มต้องเลือกว่าจะตัดออกหรือนับรวม และระบุให้ชัดเจน
- คำถามที่ควรถามเสมอ: **ผู้ใดได้รับ label ผู้ใดเป็นผู้กำหนด กำหนดเมื่อใด และผู้ใดตกหล่น**

**เวลาและการรั่วไหลของข้อมูล (time and leakage)** — ช่วงข้อมูลย้อนหลัง (look-back window: ตัวแปร) ← **index time = เวลาที่ใช้งาน** → ช่วงเวลาที่ทำนาย (prediction horizon: ผลลัพธ์)

- **Index time** คือเวลาที่ทำการทำนาย เช่น ณ วันจำหน่าย **ตัวแปรที่ใช้ได้ต้องมีอยู่ ณ เวลานั้นเท่านั้น** ข้อมูลที่บันทึกภายหลังถือเป็น**การรั่วไหลของข้อมูลจากอนาคต** (เช่น จำนวนวันที่ติดตาม หรือการวินิจฉัยเมื่อจำหน่าย)
- **ผู้ป่วยที่ปรากฏซ้ำ:** ต้องแบ่งข้อมูลตามผู้ป่วยและตัดแถวซ้ำ เพื่อมิให้บุคคลเดียวกันอยู่ทั้งในชุดฝึกและชุดทดสอบ
- **ทางลัด (shortcut):** โมเดลภาพเอกซเรย์ปอดอักเสบเรียนรู้*โรงพยาบาลและเครื่องถ่ายภาพ*แทนตัวโรค และให้ผลแย่ลงเมื่อใช้ที่โรงพยาบาลอื่น · หากได้ AUROC 0.99 ควรสงสัยการรั่วไหลของข้อมูลก่อน

## ขั้นที่ 7 — ข้อจำกัด ความปลอดภัย และเส้นทางสู่การใช้งานจริง (14:45–15:15)

**Evidence ladder:** (1) การตรวจสอบภายใน (internal validation) ด้วยข้อมูลที่กันไว้จากแหล่งเดียวกัน → (2) การตรวจสอบภายนอก (external validation) กับโรงพยาบาล ช่วงเวลา หรือประชากรใหม่ → (3) การทดลองใช้แบบเงียบหรือแบบไปข้างหน้า (silent / prospective) โดยรันกับข้อมูลจริงแต่ไม่แสดงผล → (4) การทดลองแบบสุ่มและมีกลุ่มควบคุม (RCT) ว่าช่วยให้การดูแลดีกว่าการดูแลตามปกติหรือไม่ → (5) การติดตามหลังการใช้งาน (post-market) — ต้นแบบส่วนใหญ่อยู่ในขั้นที่ 1 **ให้กลุ่มระบุขั้นถัดไปของต้นแบบของกลุ่ม**

| เครื่องมือการรายงาน | วัตถุประสงค์ |
|---|---|
| TRIPOD+AI | การรายงานการศึกษาโมเดลทำนาย (การพัฒนาหรือการตรวจสอบ) |
| PROBAST | การประเมินความเสี่ยงของอคติในการศึกษาโมเดลทำนาย |
| DECIDE-AI | การประเมินทางคลินิกระยะแรกกับผู้ใช้จริง |
| SPIRIT-AI / CONSORT-AI | โครงร่าง / รายงานการทดลองแบบสุ่มที่มี AI เป็นองค์ประกอบ |

**Software as a Medical Device (SaMD):** ระดับความเสี่ยงตาม IMDRF พิจารณาจาก**ความรุนแรงของภาวะ** (critical / serious / non-serious) ร่วมกับ**ระดับที่ผลของโมเดลมีผลต่อการตัดสินใจ** (treat or diagnose / drive / inform) แบ่งเป็นระดับ I (ต่ำสุด) ถึง IV (สูงสุด) แบบฟอร์มด้านล่างจะคำนวณระดับให้โดยอัตโนมัติ

**การติดตามหลังการใช้งาน:** เฝ้าระวัง **dataset shift** (เช่น ประชากร วิธีการตรวจทางห้องปฏิบัติการ หรือการลงรหัสที่เปลี่ยนไป) กำหนด**เกณฑ์การเตือนไว้ล่วงหน้า** วางแผนการ**ปรับเทียบใหม่ (recalibrate) ฝึกใหม่ (refit) หรือยุติการใช้ (retire)** และระบุ**ผู้มีอำนาจปิดการใช้งานโมเดล**

## รายงานของกลุ่ม (รวบรวมคำตอบจากทุกขั้น)

ให้รันเซลล์ด้านล่างเพื่อรวบรวมคำตอบทั้งหมดเป็นรายงานฉบับเดียว เซลล์จะแสดงข้อที่ยังไม่ได้กรอก (⬜) และบันทึกไฟล์ `capstone_report.md` ซึ่งใช้เป็นโครงของสไลด์ได้ ทั้งนี้ ต้องรันแบบฟอร์มของทุกขั้นก่อน

```python
# @title 📄 สร้างรายงานของกลุ่ม → capstone_report.md
# @markdown ทำเครื่องหมายเพื่อดาวน์โหลดไฟล์รายงานลงในเครื่อง (เฉพาะใน Colab)
DOWNLOAD_REPORT = False  # @param {type:"boolean"}

# หัวรายงาน: ใช้ชื่อกลุ่มจากขั้นที่ 1 (หากยังไม่ได้กรอก จะแสดงข้อความแทน)
lines, total, done = [f"# Capstone report — {REPORT.get('1. กลุ่มและโครงงาน', [('', '')])[0][1] or '(ยังไม่ได้ระบุชื่อกลุ่ม)'}", ""], 0, 0
# วนรวบรวมคำตอบของแต่ละขั้นตามลำดับใน SECTIONS
for sec in SECTIONS:
    lines.append(f"## {sec}")
    if sec not in REPORT:
        lines += ["⬜ *ยังไม่ได้รันแบบฟอร์มของขั้นนี้*", ""]
        continue
    for label, v in REPORT[sec]:
        ok = filled(v); total += 1; done += ok
        shown = ("ดำเนินการแล้ว" if v else "ยังไม่ได้ดำเนินการ") if isinstance(v, bool) else (v if ok else "…")
        lines.append(f"- {'✅' if ok else '⬜'} **{label}:** {shown}")
    lines.append("")
# สรุปจำนวนข้อที่กรอกครบ และขั้นที่ยังไม่ได้รัน
missing = [s for s in SECTIONS if s not in REPORT]
lines.insert(1, f"กรอกครบ {done} จาก {total} ข้อ" + (f" · ขั้นที่ยังไม่ได้รัน: {', '.join(missing)}" if missing else ""))
report_md = "\n".join(lines)
# บันทึกเป็นไฟล์ Markdown และแสดงในโน้ตบุ๊ก
with open("capstone_report.md", "w", encoding="utf-8") as f:
    f.write(report_md)
display(Markdown(report_md))
print("💾 บันทึกไฟล์ capstone_report.md แล้ว (ดูได้ที่แถบ Files ทางซ้ายของ Colab)")
if DOWNLOAD_REPORT:
    try:
        from google.colab import files
        files.download("capstone_report.md")
    except ImportError:
        print("การดาวน์โหลดอัตโนมัติใช้ได้เฉพาะใน Colab")
```
